```python
temperatura_interna = float(input("Digite a temperatura interna (°C): "))
temperatura_externa = float(input("Digite a temperatura externa (°C): "))
integridade = int(input("Digite a integridade estrutural (1 = íntegra / 0 = falha): "))
pressao_helio = float(input("Digite a pressão do tanque de hélio (bar): "))
pressao_combustivel = float(input("Digite a pressão do tanque de combustível (bar): "))
pressao_comburente = float(input("Digite a pressão do tanque de comburente (bar): "))

capacidade_total = float(input("Digite a capacidade total da bateria (kWh): "))
carga_atual = float(input("Digite a carga atual da bateria (%): "))
consumo_decolagem = float(input("Digite o consumo estimado na decolagem (kWh): "))
perda_energetica = float(input("Digite a perda energética (%): "))

propulsao = input("Módulo de propulsão (ok/falha): ")
navegacao = input("Módulo de navegação (ok/falha): ")
modulo_energia = input("Módulo de energia (ok/falha): ")
separacao = input("Módulo de separação (ok/falha): ")

# Cálculo de energia:
energia_inicial = capacidade_total * (carga_atual / 100)
energia_apos_consumo = energia_inicial - consumo_decolagem
perda = energia_apos_consumo * (perda_energetica / 100)
energia_restante = energia_apos_consumo - perda

# Verificação de segurança:
seguro = True

if temperatura_interna < 15 or temperatura_interna > 26.5:
    seguro = False

if temperatura_externa < 15 or temperatura_externa > 25:
    seguro = False

if integridade != 1:
    seguro = False

if carga_atual < 80 or carga_atual > 100:
    seguro = False

if energia_restante < 50:
    seguro = False

if pressao_helio < 280 or pressao_helio > 320:
    seguro = False

if pressao_combustivel < 24 or pressao_combustivel > 26:
    seguro = False

if pressao_comburente < 24 or pressao_comburente > 26:
    seguro = False

if propulsao != "ok":
    seguro = False

if navegacao != "ok":
    seguro = False

if modulo_energia != "ok":
    seguro = False

if separacao != "ok":
    seguro = False

# Resultados

print("Energia inicial:", energia_inicial, "kWh")
print("Energia após o consumo:", energia_apos_consumo, "kWh")
print("Perda energética:", perda, "kWh")
print("Energia restante:", energia_restante, "kWh")

if seguro:
    print("PRONTO PARA DECOLAR")
else:
    print("DECOLAGEM ABORTADA")
```
